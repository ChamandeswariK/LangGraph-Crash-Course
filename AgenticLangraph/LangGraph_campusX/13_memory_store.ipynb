```python
from langgraph.store.memory import InMemoryStore
```

```python
store = InMemoryStore()
```

```python
# create name space (creating folders under namespace)
namespace = ("user","u1")
```

# creating memory

```python
store.put(namespace,"1", {"data":"User like to play football"})
store.put(namespace,"2", {"data":"User like to dark mode"})
```

```python
# Another Namespace
namespace2 = ("user","u2")

# Adding Memory
store.put(namespace2,"1", {"data":"User like to play cricket"})
store.put(namespace2,"2", {"data":"User like to play hockey"})
```

# Retrieving Memory

```python
store.get(namespace,"1")
```

# Retrieving all Memory

```python
items = store.search(namespace)
for item in items:
    print(item.value)
```

# Semantic Search

```python
from langchain_ollama import OllamaEmbeddings
```

```python
embed = OllamaEmbeddings(
    model = "nomic-embed-text"
)
```

```python
store = InMemoryStore(index={'embed': embed, 'dims': 1536})
```

```python
namespace3 = ("user","u3")
```

```python
store.put(namespace3,"1",{"data":"User like to play football"})
store.put(namespace3,"2",{"data":"User like to play cricket"})
store.put(namespace3,"3",{"data":"User like pizza"})
store.put(namespace3,"4",{"data":"User like to watch movies and series"})
store.put(namespace3,"5",{"data":"User like to play video games"})
store.put(namespace3,"6",{"data":"User like to play hockey"})
store.put(namespace3,"7",{"data":"User like to learn python"})
store.put(namespace3,"8",{"data":"User like traveling"})
store.put(namespace3,"9",{"data":"User learning RAG Evals"})
store.put(namespace3,"10",{"data":"User likes real-world anologies"})
```

```python
items = store.search(namespace3, query="User like to go trip?", limit=1)

for item in items:
    print(item.value)
```

```python
items = store.search(namespace3,query = "what are the user references?", limit=2)

for item in items:
    print(item.value)
```

```python
items = store.search(namespace3, query="what is user currently learning?", limit=2)

for item in items:
    print(item.value)
```
